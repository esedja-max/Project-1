# Data exploration
Exploring the CFPB complaints used for the classifier (January-February 2026, complaints with a narrative).

Put the CFPB export (`.zip` or `.csv`) in the project's `data/` folder before running.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent          # this notebook lives in notebooks/
DATA = ROOT / "data"
SPLITS = {"train.csv", "val.csv", "test.csv"}

raw = sorted(p for p in DATA.iterdir()
             if p.suffix.lower() in {".csv", ".zip"} and p.name not in SPLITS)[0]
print("Reading:", raw.name)

df = pd.read_csv(raw, usecols=["Product", "Consumer complaint narrative"])
df = df.dropna(subset=["Consumer complaint narrative"])
df = df.rename(columns={"Product": "label", "Consumer complaint narrative": "text"})
print(df.shape)
df.head()
```

## Class balance

```python
counts = df["label"].value_counts()
print(counts)

counts.plot(kind="barh", figsize=(8, 5), title="Complaints per product")
plt.tight_layout()
plt.savefig(ROOT / "results" / "class_balance.png", dpi=150)
plt.show()
```

## Complaint length

```python
df["n_words"] = df["text"].str.split().str.len()
print(df["n_words"].describe())

df["n_words"].clip(upper=1000).hist(bins=50)
plt.title("Complaint length (words, capped at 1000)")
plt.xlabel("words")
plt.savefig(ROOT / "results" / "text_length.png", dpi=150)
plt.show()
```

## Example complaints

```python
for label in df["label"].value_counts().index[:3]:
    print("===", label)
    print(df[df["label"] == label]["text"].iloc[0][:500], "\n")
```
